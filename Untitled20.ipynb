```python
class Ave:
    def voar(self):
        print("Voando alto no céu!")

class Pato(Ave):
    pass  # Pato herda de Ave e voa perfeitamente.

# Agora você cria o Pinguim, que biologicamente é uma ave...
class Pinguim(Ave):
    def voar(self):
        # Pinguins não voam! O programador lança um erro aqui:
        raise Exception("Pinguins não sabem voar!")

# Criando e chamando o Pato
pato = Pato()
pato.voar()  # Funciona porque herda de Ave

# Criando o Pinguim
pinguim = Pinguim()
try:
    pinguim.voar()  # Vai lançar uma exceção
except Exception as e:
    print(f"Erro esperado: {e}")
```

```python
### Resumo da Filosofia de Liskov (LSP)

O princípio foca em **garantir contratos de comportamento**:
1. Uma classe filha nunca deve herdar um comportamento que ela não consegue realizar (como `Pinguim` herdando `voar`).
2. A solução de usar **Classes Intermediárias** (como `AveQueVoa`) serve para separar os comportamentos específicos sem duplicar código.
3. Qualquer subclasse deve poder substituir sua classe pai diretamente sem quebrar o sistema.
```

```python
# 1. Classe Principal
class ContaBancaria:
    def __init__(self, saldo: float):
        self.saldo = saldo

    def depositar(self, valor: float):
        self.saldo += valor

# 2. Classe Secundária (Especialização)
class ContaComSaque(ContaBancaria):
    def sacar(self, valor: float):
        self.saldo -= valor

# 3. Classes Finais
class ContaCorrente(ContaComSaque):
    pass

class ContaPoupanca(ContaComSaque):
    pass

class ContaBloqueada(ContaBancaria):
    pass  # Só deposita, nunca quebra o sistema tentando sacar!
```

```python
from abc import ABC, abstractmethod

# 1. Interfaces pequenas e específicas
class Impressora(ABC):
    @abstractmethod
    def imprimir(self, documento: str):
        pass

class Scanner(ABC):
    @abstractmethod
    def escanear(self, documento: str):
        pass

class Fax(ABC):
    @abstractmethod
    def enviar_fax(self, documento: str):
        pass

# 2. Classes concretas implementam APENAS o que usam
class ImpressoraSimples(Impressora):
    def imprimir(self, documento: str):
        print(f"Imprimindo: {documento}")

class MultifuncionalModerna(Impressora, Scanner):
    def imprimir(self, documento: str):
        print(f"Imprimindo: {documento}")

    def escanear(self, documento: str):
        print(f"Escaneando: {documento}")

# Correção: Instanciando com () e passando uma string como argumento
impressora = MultifuncionalModerna()
impressora.escanear("Contrato_Final.pdf")
```

```python
class Produto:
    def __init__(self, nome, preco):
        # Guarda os valores recebidos dentro dos atributos do próprio objeto (self)
        self.nome = nome
        self.preco = preco


class Carrinho:
    def __init__(self):
        # Inicializa o carrinho com uma lista vazia para armazenar os objetos Produto
        self.itens = []

    def adicionar(self, produto):
        # Adiciona o objeto 'produto' ao final da lista self.itens
        self.itens.append(produto)

    def remover(self, produto):
        # Verifica se o produto existe na lista antes de remover para evitar erro (ValueError)
        if produto in self.itens:
            self.itens.remove(produto)

    def total(self):
        # Percorre todos os objetos Produto guardados na lista e soma o atributo .preco de cada um
        soma = 0
        for item in self.itens:
            soma += item.preco
        return soma
```

```python
class Cliente:
    def __init__(self, cpf: str, nome: str):
        self.nome = nome
        self.cpf = cpf

class Conta:
    def __init__(self, numero: int, titular: Cliente):
        self.__saldo = 0.0 #encapsulamento
        self.numero = numero
        self.titular = titular

    def depositar(self, valor: float):
        if valor > 0:
            self.__saldo += valor

    def sacar(self, valor: float):
        if valor > 0 and valor <= self.__saldo:
            self.__saldo -= valor
            return True
        else:
            return False

    def ver_saldo(self):
        return self.__saldo

    def transferir(self, valor: float, conta_destino: 'Conta'):
        if self.sacar(valor) == True:
            conta_destino.depositar(valor)
            return True
        else:
            return False


cliente1 = Cliente('45913461886', 'Guilherme')
cliente2 = Cliente('28828554860', 'Cheila')
conta1 = Conta(101, cliente1)
conta2 = Conta(102, cliente2)

conta1.depositar(500.0)
conta1.transferir(200.0, conta2)

print(f"Saldo Conta 1: R$ {conta1.ver_saldo()}")
print(f"Saldo Conta 2: R$ {conta2.ver_saldo()}")
```
