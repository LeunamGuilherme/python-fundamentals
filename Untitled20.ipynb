````python
class Ave:
    def voar(self):
        print("Voando alto no céu!")

class Pato(Ave):
    pass  # Pato herda de Ave e voa perfeitamente.

# Agora você cria o Pinguim, que biologicamente é uma ave...
class Pinguim(Ave):
    def voar(self):
        # Pinguins não voam! O programador lança um erro aqui:
        raise Exception("Pinguins não sabem voar!")

# Criando e chamando o Pato
pato = Pato()
pato.voar()  # Funciona porque herda de Ave

# Criando o Pinguim
pinguim = Pinguim()
try:
    pinguim.voar()  # Vai lançar uma exceção
except Exception as e:
    print(f"Erro esperado: {e}")
````

### Para que serve o `self` em Python?

O `self` representa a **instância específica da classe** (ou seja, o objeto real) que está executando o método naquele momento.

#### 1. Por que precisamos dele?
Quando você define um método dentro de uma classe, o Python precisa saber de qual objeto específico você está falando. O `self` é o mecanismo que conecta o método aos dados (atributos) daquele objeto.

#### 2. Analogia do Mundo Real
Imagine que a classe é uma planta de uma casa, e os objetos criados são as casas reais construídas na rua:
* Se a planta diz: *"Pintar o quarto de azul"*, a casa precisa saber: *"Qual quarto? O desta casa (meu quarto) ou o da casa vizinha?"*
* O `self` diz: *"Pinte o quarto **desta** casa específica onde o comando está sendo executado"*.

#### 3. Exemplo Prático
Repare no exemplo abaixo como o `self` permite que cada objeto tenha seus próprios dados:

```python
class Passaro:
    def __init__(self, nome):
        # Guarda o nome dentro desta instância específica
        self.nome = nome

    def apresentar(self):
        # Usa o nome desta instância
        print(f"Olá, eu sou o {self.nome}!")

# Criando dois objetos diferentes
passaro_1 = Passaro("Paco")
passaro_2 = Passaro("Zazu")

# O Python converte 'passaro_1.apresentar()' implicitamente em: Passaro.apresentar(passaro_1)
passaro_1.apresentar()  # Saída: Olá, eu sou o Paco!
passaro_2.apresentar()  # Saída: Olá, eu sou o Zazu!
```
