```python
class Ave:
    def voar(self):
        print("Voando alto no céu!")

class Pato(Ave):
    pass  # Pato herda de Ave e voa perfeitamente.

# Agora você cria o Pinguim, que biologicamente é uma ave...
class Pinguim(Ave):
    def voar(self):
        # Pinguins não voam! O programador lança um erro aqui:
        raise Exception("Pinguins não sabem voar!")

# Criando e chamando o Pato
pato = Pato()
pato.voar()  # Funciona porque herda de Ave

# Criando o Pinguim
pinguim = Pinguim()
try:
    pinguim.voar()  # Vai lançar uma exceção
except Exception as e:
    print(f"Erro esperado: {e}")
```

```python
# 1. Classe Principal
class ContaBancaria:
    def __init__(self, saldo: float):
        self.saldo = saldo

    def depositar(self, valor: float):
        self.saldo += valor

# 2. Classe Secundária (Especialização)
class ContaComSaque(ContaBancaria):
    def sacar(self, valor: float):
        self.saldo -= valor

# 3. Classes Finais
class ContaCorrente(ContaComSaque):
    pass

class ContaPoupanca(ContaComSaque):
    pass

class ContaBloqueada(ContaBancaria):
    pass  # Só deposita, nunca quebra o sistema tentando sacar!
```
