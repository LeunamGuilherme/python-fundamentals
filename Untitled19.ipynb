```python
n = int(input())

in_count = 0
out_count = 0

for _ in range(n):
  x = int(input())
  if 10 <= x <= 20:
    in_count += 1
  else:
    out_count += 1

print(f"{in_count} in")
print(f"{out_count} out")
```

```python
N = int(input())

for i in range(2, N + 1, 2):
    square = i ** 2
    print(f"{i}^2 = {square}")
```

```python
mounth = int(input())
match mounth:
  case 1:
    print("January")
  case 2:
    print("February")
  case 3:
    print("March")
  case 4:
    print("April")
  case 5:
    print("May")
  case 6:
    print("June")
  case 7:
    print("July")
  case 8:
    print("August")
  case 9:
    print("September")
  case 10:
    print("October")
  case 11:
    print("November")
  case 12:
    print("December")
  case _:
    print("Invalid month")
```
