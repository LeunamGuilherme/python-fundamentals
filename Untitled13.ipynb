```python
# 1. Criar o dicionário inicial
student = {
    "name": "Lucas",
    "grade": 7.5
}

# 2. Adicionar uma nova chave-valor (frequência)
student["attendance"] = 85

# 3. Atualizar um valor existente (alterar a nota)
student["grade"] = 9.0

print(student)
# Saída: {'name': 'Lucas', 'grade': 9.0, 'attendance': 85}
```
