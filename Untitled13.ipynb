```python
# 1. Criar o dicionário inicial
student = {
    "name": "Lucas",
    "grade": 7.5
}

# 2. Adicionar uma nova chave-valor (frequência)
student["attendance"] = 85

# 3. Atualizar um valor existente (alterar a nota)
student["grade"] = 9.0

print(student)
# Saída: {'name': 'Lucas', 'grade': 9.0, 'attendance': 85}
```

```python
# 1. Criar o dicionário inicial
student = {
    "name": "Lucas",
    "grade": 7.5
}

# 2. Adicionar uma nova chave-valor (frequência)
student["attendance"] = 85

# 3. Atualizar um valor existente (alterar a nota)
student["grade"] = 9.0

print(student)
# Saída: {'name': 'Lucas', 'grade': 9.0, 'attendance': 85}
```

```python
student = {
    "name": "Lucas",
    "grade": 9.0,
    "attendance": 85
}

# Usamos o método .items() para pegar a chave e o valor ao mesmo tempo
for key, value in student.items():
    print(f"{key}: {value}")
```

```python
stock = {
    "laptop": 15,
    "mouse": 40,
    "keyboard": 25,
    "monitor": 8
}

# Corrigido: iteramos sobre o dicionário e verificamos o valor de cada item
for key, value in stock.items():
  if value < 20:
    print(f'Low stock: {key} ({value} units)')
```
