```python
class Stack:
    def __init__(self):
        self.items = []

    def push(self, item):
        """Adiciona um elemento ao topo da pilha"""
        self.items.append(item)

    def pop(self):
        """Remove e retorna o elemento do topo da pilha"""
        if not self.is_empty():
            return self.items.pop()
        return "Stack is empty"

    def peek(self):
        """Retorna o elemento do topo sem removê-lo"""
        if not self.is_empty():
            return self.items[-1]
        return "Stack is empty"

    def is_empty(self):
        """Verifica se a pilha está vazia"""
        return len(self.items) == 0

    def size(self):
        """Retorna o tamanho da pilha"""
        return len(self.items)


# Exemplo de teste:
my_stack = Stack()
my_stack.push(10)
my_stack.push(20)
print(my_stack.pop())  # Retorna 20 (último a entrar, primeiro a sair)
```

```python
# Lê o tempo total em segundos
N = int(input())

# 1 hora tem 3600 segundos (60 * 60)
hours = N // 3600
remaining_seconds = N % 3600

# 1 minuto tem 60 segundos
minutes = remaining_seconds // 60
seconds = remaining_seconds % 60

# Imprime no formato exato requerido: horas:minutos:segundos
print(f"{hours}:{minutes}:{seconds}")
```
