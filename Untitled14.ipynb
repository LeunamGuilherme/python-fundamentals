```python
expenses = [
    {"category": "food", "amount": 15.0},
    {"category": "transport", "amount": 10.0},
    {"category": "food", "amount": 25.0},
    {"category": "entertainment", "amount": 30.0},
    {"category": "transport", "amount": 15.0}
]

category_totals = {}

for item in expenses:
    category = item["category"]
    amount = item["amount"]

    # Se a categoria já existe no dicionário, soma ao valor existente
    if category in category_totals:
        category_totals[category] += amount
    # Se é a primeira vez que a categoria aparece, inicializa o valor
    else:
        category_totals[category] = amount

print(category_totals)
```
